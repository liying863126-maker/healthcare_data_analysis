```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/owid-covid-data.csv")
```

```python
df["date"] = pd.to_datetime(df["date"])
df["date"].min(), df["date"].max()
```

```python
df.info()
```

```python
cols = [
    "location",
    "date",
    "population",
    "new_cases",
    "new_deaths",
    "total_cases",
    "total_deaths",
    "people_vaccinated",
    "total_vaccinations",
    "people_vaccinated_per_hundred",
    "total_vaccinations_per_hundred"
]
```

```python
df_clean = df[cols].copy()
```

```python
df_clean.isna().sum()
```

```python
df_clean["new_cases"] = df_clean["new_cases"].fillna(0)
df_clean["new_deaths"] = df_clean["new_deaths"].fillna(0)
```

```python
df_clean["total_cases"] = df_clean.groupby("location")["total_cases"].ffill()
df_clean["total_deaths"] = df_clean.groupby("location")["total_deaths"].ffill()
df_clean["people_vaccinated"] = df_clean.groupby("location")["people_vaccinated"].ffill()
df_clean["total_vaccinations"] = df_clean.groupby("location")["total_vaccinations"].ffill()
df_clean["people_vaccinated_per_hundred"] = df_clean.groupby("location")["people_vaccinated_per_hundred"].ffill()
df_clean["total_vaccinations_per_hundred"] = df_clean.groupby("location")["total_vaccinations_per_hundred"].ffill()
```

```python
df_clean.isna().sum()
```

```python
df_clean.to_csv("../data/clean_data_allcountries.csv", index = False)
```
